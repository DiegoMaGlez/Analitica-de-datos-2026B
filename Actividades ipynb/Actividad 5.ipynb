Listas

Son colecciones ordenadas y mutables de elementos. Esto significa que se pueden modificar, agregar o eliminar elementos después de haberla creado. Las listas pueden contener elementos de distintos tipos de datos y permiten duplicados.

Características principales:

Se definen con corchetes "[ ]"
Son mutables: se pueden cambiar sus elementos
Mantienen el orden de inserción
Permiten elementos duplicados
Se accede a sus elementos mediante índices, comenzando en 0

Casos de uso comunes: almacenar listas de tareas, colecciones de datos que cambian con el tiempo, resultados de consultas, etc.

```python
# ===== LISTAS =====
# Colección ordenada y mutable (se puede modificar)

frutas = ["manzana", "banana", "cereza"]
frutas.append("naranja")           # Agregar elemento
frutas.remove("banana")            # Eliminar elemento
frutas[0] = "pera"                 # Modificar por índice
print("Lista:", frutas)
print("Elemento en índice 1:", frutas[1])
print("Longitud:", len(frutas))
```

Tuplas

Son colecciones ordenadas e inmutables de elementos. Una vez creada, no se puede modificar su contenido (no se pueden agregar, eliminar ni cambiar
elementos).

Características principales:

Se definen con paréntesis "( )"
Son inmutables: no se pueden modificar después de creadas
Mantienen el orden de inserción
Permiten elementos duplicados
Son más eficientes en memoria y velocidad que las listas

Casos de uso comunes: representar datos que no deben cambiar, como coordenadas, fechas, o registros fijos (por ejemplo, el nombre y la edad de una 
persona). Tambien se usan para el desempaquetado de valores en variables individuales.

```python
# ===== TUPLAS =====
# Colección ordenada e inmutable (no se puede modificar)

coordenadas = (10, 20)
persona = ("Ana", 25, "Ingeniera")

print("\nTupla:", coordenadas)
print("Nombre:", persona[0])
# persona[0] = "Luis"  # Esto daría error, las tuplas no se modifican

# Desempaquetado de tuplas
nombre, edad, profesion = persona
print(f"{nombre} tiene {edad} años y es {profesion}")
```

Diccionarios

Son colecciones de elementos organizados en pares clave-valor. Cada valor se accede a través de su clave única, en lugar de un índice numérico.

Características principales:

Se definen con llaves "{ }" en formato {clave: valor}
Son mutables: se pueden agregar, modificar o eliminar pares
Desde Python 3.7 mantienen el orden de inserción
Las claves deben ser únicas (no se permiten duplicadas)
Los valores sí pueden repetirse

Casos de uso comunes: representar objetos con atributos (como un estudiante con nombre, edad y carrera), almacenar configuraciones, o hacer búsquedas rápidas por clave.

```python
# ===== DICCIONARIOS =====
# Colección de pares clave-valor, mutable

estudiante = {
    "nombre": "Carlos",
    "edad": 22,
    "carrera": "Sistemas"
}
estudiante["edad"] = 23             # Modificar valor
estudiante["universidad"] = "UDG"   # Agregar nueva clave

print("\nDiccionario:", estudiante)
print("Carrera:", estudiante["carrera"])

for clave, valor in estudiante.items():
    print(f"{clave}: {valor}")
```

Conjuntos (Sets)

Son colecciones desordenadas de elementos únicos, es decir, no permite valores duplicados. Se basa en el concepto matemático de conjuntos y es ideal para operaciones como uniones, intersecciones y diferencias.

Características principales:

Se definen con llaves "{ }" (sin pares clave-valor)
Son mutables: se pueden agregar o eliminar elementos
No mantienen orden específico
No permiten duplicados (se eliminan automáticamente)
Soportan operaciones matemáticas de conjuntos: unión ("|"), intersección ("&"), diferencia ("-")

Casos de uso comunes: eliminar duplicados de una lista, verificar pertenencia de elementos de forma eficiente, y realizar operaciones de comparación entre colecciones.

```python
# ===== CONJUNTOS (SETS) =====
# Colección desordenada, sin elementos duplicados

numeros = {1, 2, 3, 3, 4, 4, 5}   # Los duplicados se eliminan automáticamente
print("\nConjunto:", numeros)

numeros.add(6)
numeros.discard(2)
print("Conjunto modificado:", numeros)

# Operaciones de conjuntos
a = {1, 2, 3}
b = {2, 3, 4}
print("Unión:", a | b)
print("Intersección:", a & b)
print("Diferencia:", a - b)
```
